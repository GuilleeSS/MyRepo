```python
import numpy as np
import scipy.integrate as integrate

# Definimos la función f(x) = x^2
def f1(x):
    return x**2

a = 1
b = 4
n = 3
h = (b - a) / n

x_nodes = np.linspace(a, b, n + 1)

# Cálculo de métodos numéricos
left_sum = sum(f1(x_nodes[i]) * h for i in range(n))
right_sum = sum(f1(x_nodes[i+1]) * h for i in range(n))

mid_points = [a + (i + 0.5) * h for i in range(n)]
mid_sum = sum(f1(m) * h for m in mid_points)

trap_sum = (h / 2) * (f1(a) + 2 * sum(f1(x_nodes[i]) for i in range(1, n)) + f1(b))

# Validación exacta con Scipy
exact_val, _ = integrate.quad(f1, a, b)

print(f"VERIFICATION RESULTS")
print(f"Exact Analytical Value: {exact_val:.1f}")
print(f"Left Sum:            {left_sum:.2f}  (Error: {abs(exact_val - left_sum):.2f})")
print(f"Right Sum:           {right_sum:.2f}  (Error: {abs(exact_val - right_sum):.2f})")
print(f"Midpoint Sum:        {mid_sum:.2f}  (Error: {abs(exact_val - mid_sum):.2f})")
print(f"Trapezoidal Sum:     {trap_sum:.2f}  (Error: {abs(exact_val - trap_sum):.2f})")
```

```python
import numpy as np
import scipy.integrate as integrate
import matplotlib.pyplot as plt

# Definimos la función e^(-x^2)
def f2(x):
    return np.exp(-x**2)

a_g = -1
b_g = 3

# Valor exacto preciso calculado por Scipy
exact_gauss, _ = integrate.quad(f2, a_g, b_g)

# Listas para guardar la evolución del error según los nodos (n)
n_values = [4, 10, 50, 100, 500]
print(f"ANALYSIS OF THE INTEGRAL OF exp(-x^2)")
print(f"Exact reference value  (Scipy): {exact_gauss:.8f}\n")
print(f"{'Nodes (n)':<12}{'Left-wing Method (Error)':<26}{'Trapezoidal Method (Error)'}")
print("-" * 65)

for n in n_values:
    h = (b_g - a_g) / n
    nodes = np.linspace(a_g, b_g, n + 1)

    # Izquierda
    int_left = sum(f2(nodes[i]) * h for i in range(n))
    err_left = abs(exact_gauss - int_left)

    # Trapecio
    int_trap = (h / 2) * (f2(a_g) + 2 * sum(f2(nodes[i]) for i in range(1, n)) + f2(b_g))
    err_trap = abs(exact_gauss - int_trap)

    print(f"{n:<12}{int_left:.6f} ({err_left:.2e})     {int_trap:.6f} ({err_trap:.2e})")
```
